```python
import pandas as pd


def load_split(path):
    frame = pd.read_csv(path, keep_default_na=False)
    for column in ("headline", "short_description", "category"):
        frame[column] = (
            frame[column].astype("string").str.normalize("NFKC")
            .str.replace(r"\s+", " ", regex=True).str.strip()
        )
    frame["category"] = frame["category"].str.upper()
    frame["text"] = frame["headline"] + " " + frame["short_description"]
    return frame

frames = {
    "train": load_split("data/news_train.csv"),
    "val": load_split("data/news_val.csv"),
}
```

```python
import os

from training import train_model

os.environ["HF_TOKEN"] = "hf_xxx"
```

```python
MODEL_NAME = "distilbert-base"
result = train_model(
    frames=frames,
    model_name=MODEL_NAME,
    output_dir=f"artifacts/{MODEL_NAME}",
)
```
